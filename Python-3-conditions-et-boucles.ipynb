## 1. Les booléens

Une variable **booléenne** ne prend que deux valeurs : `True` (vrai) ou `False` (faux). Elle permet de vérifier une condition, par exemple pour savoir si un.e client.e a droit à une remise. Exécutez la cellule et repérez le type de `remise_accordee`.

```python
depense_client = 150
seuil_remise = 100
remise_accordee = depense_client > seuil_remise   # vrai si la dépense dépasse le seuil
print(remise_accordee)
print(type(remise_accordee))
```

**Les opérateurs de comparaison :**

- `<` inférieur à,
- `>` supérieur à,
- `<=` inférieur ou égal à,
- `>=` supérieur ou égal à,
- `==` égal à,
- `!=` différent de.

**Important :** `==` compare deux valeurs alors que `=` donne une valeur à une variable.

```python
annee = 2026
est_annee_en_cours = (2026 == annee)   # les parenthèses ne sont pas obligatoires mais aident à lire
print(est_annee_en_cours)
```

Modifiez la deuxième ligne en `est_annee_en_cours = (2026 = annee)` et exécutez. Lisez le message d'erreur puis rétablissez la ligne.

### Combiner des conditions avec `and` et `or`

Avec `and`, les deux conditions doivent être vraies :

```python
depense_client = 150
eligible = (depense_client > 100) and (depense_client < 200)
print("Remise et livraison gratuite ?", eligible)
```

Avec `or`, une seule suffit :

```python
eligible = (depense_client > 100) or (depense_client == 50)
print("Remise ou offre spéciale ?", eligible)
```

### Comparer des textes

Les comparaisons de textes servent à classer ou filtrer des clients, des produits ou des comportements :

```python
segment = 'Premium'

est_premium = ('premium' == segment.lower())   # comparaison sans tenir compte des majuscules
print("Client.e premium :", est_premium)
```

La méthode `lower()` met un texte en minuscules. On l'appelle avec un point après le nom de la variable. Cette notation sera expliquée dans Python 5.

```python
segment_minuscules = segment.lower()
print("Après lower() :", segment_minuscules)

est_premium = 'premium' == segment_minuscules
print(segment_minuscules, est_premium)

# Il n'est pas nécessaire de garder le texte en minuscules dans une variable
est_premium = 'premium' == segment.lower()
print(segment.lower(), est_premium)
```

## 2. Les conditions : `if` et `else`

Une condition exécute un bloc de code si elle est vraie et un autre si elle est fausse. On peut penser à un algorithme qui ajuste un prix selon le comportement de la clientèle.

**Important :** Python impose un **décalage** (quatre espaces) des lignes qui dépendent de la condition. C'est ce décalage qui indique où commence et où finit chaque bloc.

```python
prix = 100

if prix > 50:
    print("Le prix dépasse le seuil de remise.")
    nouveau_prix = prix * 0.9   # remise de 10 %
    print("Prix après remise :", nouveau_prix, "\n")
else:
    print("Prix trop bas pour une remise.\n")

print("Cette ligne s'exécute dans tous les cas.")
```

### Un autre exemple : les tailles de produit

```python
taille = 'moyenne'
if (taille == 'petite') or (taille == 'moyenne') or (taille == 'grande'):
    print("Une taille standard a été choisie.\n")

# else est facultatif
print("Cette ligne s'exécute dans tous les cas.")
```

## 3. Les boucles

Une **boucle** répète un bloc de code. La boucle `for` le répète un nombre de fois fixé à l'avance. Les boucles servent par exemple à appliquer le même calcul à chaque année d'une série.

### Répéter un nombre de fois donné

```python
# Répéter trois fois et afficher la valeur de i
for i in range(3):
    print("Passage :", i)
```

### La fonction `range()`

`range()` produit une suite de nombres entiers :

- `range(3)` donne 0, 1, 2 (Python commence à compter à 0),
- `range(2, 7)` donne 2, 3, 4, 5, 6 (le dernier nombre est exclu),
- `range(0, 10, 2)` donne 0, 2, 4, 6, 8 (le troisième nombre est le pas).

```python
# range est un type Python, comme float ou str
ma_suite = range(5)
print(type(ma_suite))

# Afficher directement la suite ne montre pas ses éléments
print(ma_suite)

# Pour les voir, on la transforme en liste
print(list(range(0, 10, 2)))
```

### Parcourir une liste

Une boucle peut parcourir les éléments d'une liste :

```python
indicateurs = ['Utilisateur.rices', "Chiffre d'affaires", 'Transactions', 'Publicités']

for indicateur in indicateurs:
    print("Suivi :", indicateur)
```

### Parcourir un texte

On peut aussi parcourir un texte, caractère par caractère :

```python
categorie = 'eCommerce'

for lettre in categorie:
    print(lettre)
```

Les suites `range`, les listes et les textes sont des objets **itérables** : ils savent fournir leur élément suivant. On les retrouvera souvent.
