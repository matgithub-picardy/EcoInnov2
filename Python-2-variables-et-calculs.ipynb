## 2. Les calculs

````python
# Multiplications
taux_croissance = 2 * 3
utilisateurs_actifs = 2*3          # les espaces ne changent rien au calcul
engagement_mensuel = 2     *     3

print("Taux de croissance, utilisateur.rices actif.ves et engagement mensuel :")
print(taux_croissance, utilisateurs_actifs, engagement_mensuel)

# Multiplier deux indicateurs
engagement_total = taux_croissance * utilisateurs_actifs
print("Engagement total (croissance x utilisateur.rices actif.ves) :")
print(engagement_total)
````

Ici, `print()` affiche aussi des messages placés entre guillemets. Ce sont des chaînes de caractères (partie 4).

````python
# Divisions
revenu_moyen_par_utilisateur = 10 / 2
revenu_publicitaire_par_clic = 20 / 5
taux_abonnement = 6 / 4

print("Revenu moyen par utilisateur.rice, revenu par clic et taux d'abonnement :")
print(revenu_moyen_par_utilisateur, revenu_publicitaire_par_clic, taux_abonnement)
````

La division donne des nombres à virgule, même quand le résultat tombe juste (`5.0`). La multiplication de deux entiers donnait un entier.

**Chaque objet Python a un type** : entier (`int`), nombre à virgule (`float`), chaîne de caractères (`str`), liste (`list`), etc. La fonction `type()` indique le type d'une variable.

````python
print("revenu_moyen_par_utilisateur vaut", revenu_moyen_par_utilisateur,
      "et son type est", type(revenu_moyen_par_utilisateur))

print("\n")   # \n produit un saut de ligne

print("taux_croissance vaut", taux_croissance, "et son type est", type(taux_croissance))
````

````python
# Puissance : on écrit ** et non ^
utilisateurs = 2**3   # le nombre double à chacune des trois périodes

print("Nombre d'utilisateur.rices après trois doublements :", utilisateurs)
````

````python
# Cette cellule produit volontairement une erreur : lisez le message.
# Le logarithme népérien sert souvent pour des taux de croissance
utilisateurs_numeriques = 10
log_utilisateurs = log(utilisateurs_numeriques)
````

Le logarithme ne fait pas partie des fonctions intégrées de Python. On peut en revanche ajouter des **bibliothèques** (en anglais *packages*) qui apportent de nouvelles fonctions. La bibliothèque `numpy` contient de nombreuses fonctions numériques, dont `log`. Il faut d'abord l'importer.

````python
# Importer numpy sous le nom court np
import numpy as np

# Pour utiliser une fonction de numpy, on écrit np. suivi de son nom
log_utilisateurs = np.log(utilisateurs_numeriques)
print(log_utilisateurs)

# L'exponentielle est la fonction inverse du logarithme
retour = np.exp(log_utilisateurs)
print(retour)
````

## 5. L'affichage mis en forme

Supposons que l'on veuille écrire que le chiffre d'affaires de Google était de 29,32 milliards de dollars en 2010 et de 182,53 milliards en 2020. On peut l'écrire directement :

````python
print("Chiffre d'affaires de Google en 2010 et en 2020 : 29.32 et 182.53 milliards de dollars")
````

Si ces nombres sont calculés par le programme, il faut les insérer automatiquement dans le texte. La méthode `format` le permet :

````python
print("Chiffre d'affaires de Google en 2010 et en 2020 : {0:6.2f} et {1:6.2f} milliards de dollars".format(29.32, 182.53))
````

Le texte contient deux emplacements entre accolades `{}`. Ils sont remplacés par les valeurs données à `format`. Le chiffre avant les deux-points indique la position de la valeur : `{0}` pour la première, `{1}` pour la seconde. Après les deux-points vient le format :

- `6` : la largeur totale réservée au nombre, point décimal compris,
- `.2` : deux décimales,
- `f` : un nombre à virgule fixe.

Pour afficher une accolade, on la double : `{{` ou `}}`. On peut changer l'ordre des valeurs :

````python
print("Chiffre d'affaires de Google en 2010 et en 2020 : {1:6.2f} et {0:6.2f} milliards de dollars".format(182.53, 29.32))
````

On peut aussi nommer les emplacements pour ne pas se soucier de l'ordre :

````python
print("Chiffre d'affaires de Google en 2010 et en 2020 : {ca_2010:6.2f} et {ca_2020:6.2f} milliards de dollars".format(ca_2010=29.32, ca_2020=182.53))
````

Depuis Python 3.6, les **f-strings** font la même chose plus simplement. On place un `f` devant le texte et le nom des variables entre accolades :

````python
ca_2010 = 29.32
ca_2020 = 182.53
print(f"Chiffre d'affaires de Google en 2010 et en 2020 : {ca_2010:6.2f} et {ca_2020:6.2f} milliards de dollars")
````

C'est la forme la plus lisible. C'est celle que l'on utilise dans les TD.
