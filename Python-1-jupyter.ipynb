## 1. Ce qu'est un notebook Jupyter

On peut écrire du code Python dans un fichier texte, l'enregistrer puis le faire exécuter par un programme appelé **interpréteur**. Le code et les résultats sont alors séparés. Un **notebook Jupyter** réunit les deux :

- des explications dans des cellules de **texte** (format Markdown) : ce paragraphe en est une,
- du code et ses résultats dans des cellules de **code** : celle qui suit en est une, marquée `[ ]:` à gauche.

**Tâche 1.**

1. Cliquez dans la cellule de code suivante.
2. Tapez `33*2`.
3. Exécutez la cellule avec **Maj + Entrée** ou avec le bouton **Exécuter** (triangle ▶) de la barre d'outils.

Le résultat (`66`) s'affiche juste sous la cellule. C'est le principe de Jupyter : écrire du code, l'exécuter, voir le résultat aussitôt.

`````python
33*2
`````

## 2. Travailler avec les notebooks

### Ouvrir un notebook

1. Le **panneau de gauche** affiche les fichiers et les dossiers.
2. Les notebooks sont les fichiers qui se terminent par **`.ipynb`**.
3. Double-cliquez sur un nom pour l'ouvrir dans un nouvel onglet.

### Créer un notebook

- bouton **+** en haut du panneau de gauche, **Notebook** et **Python 3 (ipykernel)**,
- ou menu **File**, **New**, **Notebook** et **Python 3 (ipykernel)**.

### Enregistrer un notebook

- icône de disquette dans la barre d'outils,
- menu **File**, **Save Notebook** (ou **Save Notebook As...** pour changer de nom),
- raccourci **Ctrl + S** (Windows et Linux) ou **Cmd + S** (Mac).

**Attention sur MyBinder.** Les notebooks y sont **temporaires** : vos modifications disparaissent quand la session se ferme. Pour garder votre travail, **téléchargez le notebook** (menu **File**, **Download**) avant de fermer l'onglet.

### Fermer un notebook

Fermez l'onglet du notebook (la petite croix). Ne fermez pas la fenêtre du navigateur si vous voulez continuer à travailler sur MyBinder.
