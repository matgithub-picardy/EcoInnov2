## 1. Les listes

Une **liste** est une collection ordonnée et modifiable. On l'écrit entre crochets :

```python
# Une liste de nombres (des montants d'achat, par exemple)
montants = [5, 20, 15, 10]

# Une liste de textes (des noms de plateformes)
plateformes = ['Facebook', 'Amazon', 'Google']

print("Montants :", montants)
print("Plateformes :", plateformes)
```

Une liste peut mélanger des types, par exemple des identifiants et des actions :

```python
liste_mixte = [101, 'recherche', 205, 'achat']

print("Liste mixte :", liste_mixte)
print("Type du premier élément :", type(liste_mixte[0]))
print("Type du deuxième élément :", type(liste_mixte[1]))
```

**À retenir :** Python numérote les éléments d'une liste **à partir de 0**. Le premier élément a l'indice 0, le deuxième l'indice 1.

```python
print("Premier élément :", liste_mixte[0])    # 101
print("Deuxième élément :", liste_mixte[1])   # 'recherche'
```

### Modifier une liste

On peut réunir des listes avec `+` et ajouter un élément avec `append()` :

```python
# Réunir deux listes
liste_reunie = liste_mixte + ['Instagram', 30]
print("Liste réunie :", liste_reunie)

# Ajouter un élément à la fin
liste_mixte.append('terminé')
print("Liste mixte complétée :", liste_mixte)
```

Une liste peut contenir d'autres listes :

```python
# Un profil : identifiant, prénom et une sous-liste (statut, nombre d'actions)
profil = [101, 'Jean', ['actif', 50]]

print("Profil complet :", profil)
print("Statut :", profil[2][0])   # 'actif'
```

L'opérateur `+` reconnaît les objets sur lesquels il travaille : il additionne des nombres, colle des textes et réunit des listes. Il ne peut pas additionner un nombre et un texte.

### Les listes en compréhension

Une **liste en compréhension** construit une liste en une seule ligne à partir d'une autre. Tout ce qu'elle fait peut se faire avec une boucle `for`, mais elle est plus courte et très répandue.

**Exemple : harmoniser des catégories.** Dans des données réelles, une même catégorie apparaît souvent sous plusieurs formes ('Abonné', 'abonné'). On met tout en minuscules pour pouvoir comparer.

Avec une boucle :

```python
types_utilisateurs = ['Abonne', 'abonne', 'Gratuit', 'invite', 'abonne']

# Liste vide qui recevra les données nettoyées
types_nettoyes = []
for t in types_utilisateurs:
    types_nettoyes.append(t.lower())

print(types_nettoyes)
```

Avec une liste en compréhension :

```python
types_nettoyes_2 = [t.lower() for t in types_utilisateurs]
print(types_nettoyes_2)
```

Autre exemple, mettre au carré des nombres d'interactions :

```python
interactions = [2, 3, 5]
carres = [n**2 for n in interactions]
print(carres)
```

On peut ajouter une condition pour filtrer, par exemple ne garder que les abonné.es :

```python
abonnes = [t.upper() for t in types_nettoyes if t == 'abonne']
print(abonnes)
```

## 2. Les tuples

Un **tuple** ressemble à une liste mais ne peut plus être modifié une fois créé. On l'utilise pour des données qui doivent rester fixes (les paramètres d'un modèle par exemple). On l'écrit entre parenthèses :

```python
tuple_nombres = (2, 3, 5, 8)
tuple_textes = ('prix', 'élasticité', 'demande')

print("Type de tuple_nombres :", type(tuple_nombres))
print("tuple_nombres :", tuple_nombres, "\n")
print("Type de tuple_textes :", type(tuple_textes))
print("tuple_textes :", tuple_textes)
```

Comparons une liste et un tuple quand on essaie de modifier un élément :

```python
# Cette cellule produit volontairement une erreur : lisez le message.
# Une liste se modifie
liste_nombres = [2, 3, 5, 8]
liste_nombres[1] = 1000
print(liste_nombres)

# Un tuple ne se modifie pas
tuple_nombres[1] = 1000
```

## 3. Les dictionnaires

Un **dictionnaire** associe des **clés** à des **valeurs**. Un vrai dictionnaire associe de même un mot à sa définition. Les clés sont uniques, les valeurs peuvent se répéter. On l'écrit entre accolades. Associons des formules d'abonnement à leur prix :

```python
prix_abonnements = {'Basique': 5.99, 'Pro': 15.99, 'Premium': 29.99}

print(type(prix_abonnements))
print(prix_abonnements)
prix_abonnements.keys()
```

'Basique', 'Pro' et 'Premium' sont les clés. Les prix sont les valeurs. On accède à une valeur par sa clé :

```python
print(prix_abonnements['Pro'])   # affiche 15.99
```

### Une erreur fréquente

On ne peut pas chercher une clé à partir d'une valeur :

```python
# Cette cellule produit volontairement une erreur : lisez le message.
# Un dictionnaire se consulte par ses clés, pas par ses valeurs
print(prix_abonnements[15.99])
```

Python renvoie une `KeyError` : il cherche une clé 15.99 qui n'existe pas.

### Ajouter ou modifier une entrée

Un dictionnaire est modifiable :

```python
# Ajouter une formule
prix_abonnements['Entreprise'] = 49.99
print(prix_abonnements)

# Modifier le prix de la formule Basique
prix_abonnements['Basique'] = 6.99
print(prix_abonnements)
```
