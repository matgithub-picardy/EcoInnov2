## 1. Le découpage

Le découpage extrait une partie d'une liste, d'un tuple ou d'un texte. On l'écrit entre crochets :

```python
donnees[debut:fin:pas]
```

- `debut` : l'indice de départ (inclus),
- `fin` : l'indice d'arrivée (**exclu**),
- `pas` : l'écart entre deux éléments retenus (1 par défaut).

Le découpage sert par exemple à garder les dernières années d'une série ou une observation sur deux.

````python
visites = [105, 204, 304, 404, 505]

print(visites, "  # liste de départ")
print(visites[0:2])     # les deux premiers éléments
print(visites[0:4:1])   # les quatre premiers, avec un pas explicite de 1
print(visites[0:5:2])   # un élément sur deux
````

Si on omet le début ou la fin, le découpage part du début ou va jusqu'à la fin :

````python
print(visites, "  # liste de départ")
print(visites[2:])    # du troisième au dernier
print(visites[:4])    # les quatre premiers
````

### Couper une liste en deux

````python
premiere_moitie = visites[:3]
seconde_moitie = visites[3:]
print(premiere_moitie, "\n", seconde_moitie)
````

### Les indices négatifs

Un indice négatif compte à partir de la fin : `-1` désigne le dernier élément. C'est utile pour garder les observations les plus récentes.

````python
print(visites, "  # liste de départ")
print(visites[:-1])     # tout sauf le dernier
print(visites[:-2])     # tout sauf les deux derniers
print(visites[-4:-2])   # du quatrième en partant de la fin jusqu'au troisième en partant de la fin
````

### Un pas négatif

Un pas négatif parcourt la liste à l'envers :

````python
print(visites[::-1])     # la liste à l'envers
print(visites[4:1:-1])   # de l'indice 4 à l'indice 2, en reculant
````

Le découpage fonctionne aussi sur les textes :

````python
mot_cle = '#numerique'
print(mot_cle[:1])     # le premier caractère : '#'
print(mot_cle[1:])     # le mot sans le #
print(mot_cle[::-1])   # à l'envers
````

## 2. Les fonctions

Une **fonction** regroupe des instructions sous un nom pour les réutiliser. On écrit le code une fois et on l'appelle autant de fois que nécessaire. On la définit avec `def` :

````python
def portee_estimee(interactions):
    """
    Estime le nombre de personnes atteintes par une publication à partir de son
    nombre d'interactions, en supposant que 5 % des personnes atteintes réagissent.
    """
    portee = interactions / 0.05
    return portee   # la valeur renvoyée par la fonction
````

Une fois la cellule exécutée, la fonction `portee_estimee` est disponible. La commande `whos` liste les objets créés jusqu'ici, dont cette fonction.

````python
whos
````

Utilisons la fonction :

````python
interactions_publication = 250   # mentions j'aime, commentaires et partages
estimation = portee_estimee(interactions_publication)
print("La publication a atteint environ", estimation, "personnes.")
````

Une bonne fonction prévoit les erreurs possibles. Observez ce qui se passe si on lui donne un texte :

````python
# Cette cellule produit volontairement une erreur : lisez le message.
interactions_publication = '250'   # un texte au lieu d'un nombre
estimation = portee_estimee(interactions_publication)
````

````python
def portee_estimee_v2(interactions):
    """
    Même calcul, avec une vérification du type de l'entrée.
    """
    if isinstance(interactions, (int, float)):
        return interactions / 0.05
    else:
        print("Erreur : portee_estimee_v2 attend un nombre.")
        return None
````

````python
estimation = portee_estimee_v2('250')
print("Estimation :", estimation)
````

Tout l'art consiste à trouver l'équilibre entre le temps passé à prévoir les erreurs et la robustesse dont on a besoin.

Une fonction peut prendre plusieurs entrées :

````python
def formater_profil(prenom, pseudo, nom):
    """
    Met en forme le nom et le pseudo d'une personne pour l'affichage d'un profil.
    """
    return prenom.title() + " " + nom.title() + " (" + pseudo.lower() + ")"
````

````python
profil = formater_profil('ALex', '@aLexCode', 'martin')
print(profil)
````

### Affecter plusieurs variables à la fois

On peut donner une valeur à plusieurs variables en une seule ligne :

````python
identifiant, nombre_publications = 12345, 98
print(identifiant, nombre_publications)
````

Cela permet d'échanger deux valeurs en une ligne :

````python
id_a = 1001
id_b = 2002
print("Avant l'échange : id_a =", id_a, " id_b =", id_b)
id_a, id_b = id_b, id_a
print("Après l'échange : id_a =", id_a, " id_b =", id_b)
````

Une fonction peut ainsi renvoyer plusieurs résultats. Celle-ci calcule le taux de rebond et le taux de conversion d'une page web :

````python
def indicateurs_web(visites, rebonds, conversions):
    """
    Taux de rebond (visites qui quittent la page aussitôt) et taux de conversion
    (visites qui aboutissent à un achat), en pourcentage des visites.
    """
    taux_rebond = rebonds / visites * 100
    taux_conversion = conversions / visites * 100
    return taux_rebond, taux_conversion

rebond, conversion = indicateurs_web(500, 150, 50)
print(f"Taux de rebond : {rebond:.2f} % | Taux de conversion : {conversion:.2f} %")
````

## 3. Les objets et la complétion avec Tab

En Python, **tout est objet** : les variables, les listes, les textes, les fonctions. Chaque objet dispose de **méthodes**, des fonctions qui lui sont attachées et qui dépendent de son type. On les appelle avec un point : `objet.methode()`. Nous avons déjà utilisé `lower()`, `upper()` et `title()` sur des textes, `append()` sur des listes.

Pour une liste par exemple :

````python
liste_1 = ['a', 'c', 'b']
print(liste_1)
````

````python
liste_1.sort()   # la méthode sort() du type list trie la liste
print(liste_1)
````

### Trouver les méthodes d'un objet

Une recherche sur Internet aide, mais Jupyter offre plus rapide : la **complétion avec la touche Tab**. Tapez `liste_1.` dans la cellule suivante puis appuyez sur Tab.

La liste des méthodes disponibles s'affiche. Essayons `reverse()` :

````python
liste_1.reverse()
print(liste_1)
````

La touche Tab complète aussi les noms de variables : tapez `lis` puis Tab. Cela évite les fautes de frappe.
