Une **variable** garde une valeur sous un nom. On peut ensuite la réutiliser.

```python
rd_france = 2.18       # dépenses de R&D en % du PIB, France, 2023
objectif_ue = 3.0      # objectif européen

ecart = objectif_ue - rd_france
print("Écart à l'objectif :", round(ecart, 2), "point de PIB")
```

Une **liste** rassemble plusieurs valeurs. Une **boucle** répète une instruction pour chaque élément.

```python
pays = ["France", "Allemagne", "Suède"]
intensites = [2.18, 3.15, 3.60]   # année 2023

for nom, valeur in zip(pays, intensites):
    print(nom, ":", valeur, "% du PIB")
```

## 1. Charger la base (Essentiel)

La base `rd-pays-2000-2023.csv` rassemble trois indicateurs pour 19 pays de 2000 à 2023 :

- `rd_pib` : dépenses intérieures de R&D en pourcentage du PIB, c'est-à-dire l'intensité de R&D (cf. Chapitre 3),
- `chercheurs_par_million` : nombre de chercheur.ses pour un million d'habitant.es,
- `pib_par_habitant` : PIB par habitant en parité de pouvoir d'achat (dollars internationaux constants de 2021).

```python
# Lire le fichier : chaque ligne du fichier devient une ligne du tableau
base = pd.read_csv("donnees/rd-pays-2000-2023.csv")

# Afficher les cinq premières lignes
base.head()
```

## 4. Un premier graphique (Essentiel)

```python
# Choisir les pays à comparer (vous pouvez modifier la liste)
choix = ["France", "Allemagne", "Corée du Sud", "États-Unis"]
selection = base[base["pays"].isin(choix)]

for nom in choix:
    donnees_pays = selection[selection["pays"] == nom]
    plt.plot(donnees_pays["annee"], donnees_pays["rd_pib"], label=nom)

plt.title("Dépenses de R&D en % du PIB")
plt.xlabel("Année")
plt.ylabel("R&D (% du PIB)")
plt.legend()
plt.show()
```

## 5. Comparer et interpréter (Essentiel, à distance)

**Tâche A.** Intensité de R&D moyenne de chaque pays sur la période.

```python
moyennes = base.groupby("pays")["rd_pib"].mean().sort_values(ascending=False)
moyennes
```

**Tâche B (Pour aller plus loin).** On cherche à savoir si les pays qui consacrent le plus à la R&D sont aussi les plus riches.

```python
annee = 2021
coupe = base[base["annee"] == annee]

plt.scatter(coupe["pib_par_habitant"], coupe["rd_pib"])
for _, ligne in coupe.iterrows():
    plt.annotate(ligne["code"], (ligne["pib_par_habitant"], ligne["rd_pib"]))
plt.title(f"Intensité de R&D et PIB par habitant en {annee}")
plt.xlabel("PIB par habitant (dollars PPA de 2021)")
plt.ylabel("R&D (% du PIB)")
plt.show()

# Corrélation entre les deux indicateurs (de -1 à 1)
print(round(coupe["pib_par_habitant"].corr(coupe["rd_pib"]), 2))
```
